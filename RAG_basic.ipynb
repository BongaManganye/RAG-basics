```python
from langchain_community.document_loaders import PyPDFLoader

PDF_PATH = "telecom_guide.pdf"

loader = PyPDFLoader(PDF_PATH)
pages = loader.load()

print(f"Loaded {len(pages)} pages from the PDF.")
print("\n --- First page preview (first 500 chars) ---")
print(pages[0].page_content[:500])
```

```python
print(pages[1].page_content[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=600,
    chunk_overlap=100,
    separators=["\n\n", "\n", ",", " "]
)

chunks = splitter.split_documents(pages)

len(chunks)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
vector_store = Chroma.from_documents(chunks, embeddings)

print(f"Vector store ready. {vector_store._collection.count()} vectors stored.")
```

```python
retriever = vector_store.as_retriever(search_kwargs={"k": 3})

test_query = "What is  VolTE and how does ot improve call quality"
retrieved = retriever.invoke(test_query)

for i, doc in enumerate(retrieved, 1):
    print(f" --- Chunk [i] ---")
    print(doc.page_content[:300])
    print()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_groq import ChatGroq

# 1. Helper function
def format_docs(docs):
    return "\n\n---\n\n".join(doc.page_content for doc in docs)

# 2. System prompt
SYSTEM_PROMPT = """You are a helpful telecom assistant.
Answer the question using ONLY the context provided below.
If the context does not contain enough information, say so clearly.

Context:
{context}"""

# 3. Prompt setup
prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT),
    ("human", "{question}")
])

# 4. LLM setup
llm = ChatGroq(
    model="qwen/qwen3-32b",
    temperature=0,
    reasoning_format="parsed"
)

# 5. Output parser
parser = StrOutputParser()

# 6. Inputs mapping
inputs = {
    "context": retriever | format_docs,
    "question": RunnablePassthrough()
}

# 7. Final chain assembly
chain = inputs | prompt | llm | parser

print("RAG chain assembled.")
```

```python
question = "How does international roaming work and what charges should I expect?"

print(f"Q: {question}\n")
print("A:", chain.invoke(question))
```
